# Programa de calificaciones
Este programa calcula el promedio de 3 calificaciones y evalúa si un alumno **aprobó** o **no aprobó**.

## Definición de variables

```python
calificacion1 = 0.0
calificacion2 = 0.0
calificacion3 = 0.0
promedio = 0.0
```

# Entrada de datos

```python
calificacion1 = float(input("Ingrese la calificación 1"))
calificacion2 = float(input("Ingrese la calificación 2"))
calificacion3 = float(input("Ingrese la calificacion 3"))
```

# Proceso: Cálculo de promedio

```python
promedio = (calificacion1 + calificacion2 + calificacion3) / 3
```

# Comparación con una estrucutra selectiva doble

```python
if promedio >= 7.0:
    print("Aprobado")
else:
    print("No aprobado")
```
